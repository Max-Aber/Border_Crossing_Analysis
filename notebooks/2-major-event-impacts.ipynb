```python
import sys
import pandas as pd
import importlib
sys.path.append('..')

import scripts.plot_utils as plot
importlib.reload(plot)
```

```python
df_total_monthly = pd.read_csv('../data/processed/monthly_crossings_people.csv')
df_total_monthly['Date'] = pd.to_datetime(df_total_monthly['Date'])
fig, ax = plot.plot_time_series(df_total_monthly, "Date", "Total_Crossings",'Total Crossings Over Time', 'Date', 'People', show=False)

# Highlight significant periods
ax.axvspan(pd.to_datetime('2001-09-01'), pd.to_datetime('2001-9-11'), 
           alpha=0.15, color='blue', label='911')
ax.axvspan(pd.to_datetime('2008-01-01'), pd.to_datetime('2009-12-31'), 
           alpha=0.15, color='orange', label='Financial Crisis')
ax.axvspan(pd.to_datetime('2017-01-20'), pd.to_datetime('2021-01-30'), 
           alpha=0.15, color='white', label='1st Trump Term')
ax.axvspan(pd.to_datetime('2020-02-01'), pd.to_datetime('2022-06-30'), 
           alpha=0.15, color='red', label='COVID-19')
ax.axvspan(pd.to_datetime('2025-01-20'), pd.to_datetime('2025-05-20'), 
           alpha=0.15, color='white', label='2nd Trump Term')


ax.legend()
fig.show()
```
